```python
# Εισάγουμε απαραίτητες βιβλιοθήκες
import datetime

# Ορίζουμε μια κλάση για την αναπαράσταση ενός περιστατικού ασφάλειας
class Incident:
    def __init__(self, title, description, severity):
        # Τίτλος του περιστατικού
        self.title = title
        # Περιγραφή του περιστατικού
        self.description = description
        # Βαθμός σοβαρότητας (1=χαμηλή, 5=πολύ σοβαρή)
        self.severity = severity
        # Ημερομηνία και ώρα καταγραφής
        self.timestamp = datetime.datetime.now()
        # Κατάσταση (ανοιχτό/κλειστό)
        self.status = "Ανοιχτό"
        # Λίστα ενεργειών που έγιναν για το περιστατικό
        self.actions = []

    # Προσθήκη ενέργειας αντιμετώπισης
    def add_action(self, action):
        # Καταγραφή ενέργειας και ώρας που έγινε
        entry = f"{datetime.datetime.now()}: {action}"
        self.actions.append(entry)

    # Κλείσιμο του περιστατικού
    def close(self):
        self.status = "Κλειστό"

    # Αναπαράσταση του περιστατικού με string
    def __str__(self):
        return (
            f"--- Περιστατικό ---\n"
            f"Τίτλος: {self.title}\n"
            f"Περιγραφή: {self.description}\n"
            f"Σοβαρότητα: {self.severity}/5\n"
            f"Κατάσταση: {self.status}\n"
            f"Ώρα καταγραφής: {self.timestamp}\n"
            f"Ενέργειες:\n" + "\n".join(self.actions) + "\n"
        )


# Βασικό σύστημα διαχείρισης περιστατικών
class IncidentResponseSystem:
    def __init__(self):
        # Λίστα με όλα τα περιστατικά
        self.incidents = []

    # Δημιουργία νέου περιστατικού
    def report_incident(self, title, description, severity):
        incident = Incident(title, description, severity)
        self.incidents.append(incident)
        print("📌 Νέο περιστατικό καταγράφηκε επιτυχώς.")

    # Προβολή όλων των περιστατικών
    def list_incidents(self):
        if not self.incidents:
            print("⚠️ Δεν υπάρχουν καταγεγραμμένα περιστατικά.")
            return
        for idx, inc in enumerate(self.incidents, 1):
            print(f"[{idx}] {inc.title} - Σοβαρότητα: {inc.severity} - Κατάσταση: {inc.status}")

    # Προβολή αναλυτικών στοιχείων για συγκεκριμένο περιστατικό
    def view_incident(self, index):
        if 0 <= index < len(self.incidents):
            print(self.incidents[index])
        else:
            print("❌ Μη έγκυρος αριθμός περιστατικού.")

    # Προσθήκη ενέργειας σε περιστατικό
    def add_action_to_incident(self, index, action):
        if 0 <= index < len(self.incidents):
            self.incidents[index].add_action(action)
            print("✅ Η ενέργεια προστέθηκε.")
        else:
            print("❌ Μη έγκυρος αριθμός περιστατικού.")

    # Κλείσιμο περιστατικού
    def close_incident(self, index):
        if 0 <= index < len(self.incidents):
            self.incidents[index].close()
            print("🔒 Το περιστατικό έκλεισε.")
        else:
            print("❌ Μη έγκυρος αριθμός περιστατικού.")


# Κεντρικό μενού για δοκιμή του συστήματος
def main():
    system = IncidentResponseSystem()

    while True:
        print("\n📋 Μενού Αντιμετώπισης Περιστατικών:")
        print("1. Καταγραφή νέου περιστατικού")
        print("2. Προβολή όλων των περιστατικών")
        print("3. Προβολή λεπτομερειών περιστατικού")
        print("4. Προσθήκη ενέργειας σε περιστατικό")
        print("5. Κλείσιμο περιστατικού")
        print("6. Έξοδος")

        choice = input("Επιλογή: ")

        if choice == "1":
            title = input("Τίτλος περιστατικού: ")
            description = input("Περιγραφή: ")
            severity = int(input("Σοβαρότητα (1-5): "))
            system.report_incident(title, description, severity)
        elif choice == "2":
            system.list_incidents()
        elif choice == "3":
            idx = int(input("Αριθμός περιστατικού: ")) - 1
            system.view_incident(idx)
        elif choice == "4":
            idx = int(input("Αριθμός περιστατικού: ")) - 1
            action = input("Περιγραφή ενέργειας: ")
            system.add_action_to_incident(idx, action)
        elif choice == "5":
            idx = int(input("Αριθμός περιστατικού: ")) - 1
            system.close_incident(idx)
        elif choice == "6":
            print("👋 Έξοδος από το σύστημα.")
            break
        else:
            print("❗ Μη έγκυρη επιλογή.")

# Εκκίνηση του προγράμματος
if __name__ == "__main__":
    main()
```
